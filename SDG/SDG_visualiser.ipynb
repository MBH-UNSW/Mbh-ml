```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# loading the generated synthetic data
# lowk have to change file location sometimes
df = pd.read_csv('tah_synthetic_healthy.csv')
df.head()
```

```python
t = df['timestamp']
pressure_systemic = df['pressure_systemic']
pressure_pulmonary = df['pressure_pulmonary']
flow = df['flow_rate']
temp = df['temperature']
```

```python
# only plotting first 1000 points (100 seconds) otherwise its just a solid block
n = 4000

fig, axes = plt.subplots(2, 2, figsize=(12, 6))
fig.suptitle('TAH Synthetic Telemetry - First 100 Seconds')

axes[0,0].plot(t[:n], pressure_systemic[:n])
axes[0,0].set_title('Systemic Pressure')
axes[0,0].set_ylabel('mmHg')

axes[0,1].plot(t[:n], pressure_pulmonary[:n])
axes[0,1].set_title('Pulmonary Pressure')
axes[0,1].set_ylabel('mmHg')

axes[1,0].plot(t[:n], flow[:n])
axes[1,0].set_title('Flow Rate')
axes[1,0].set_ylabel('L/min')

axes[1,1].plot(t[:n], temp[:n])
axes[1,1].set_title('Temperature')
axes[1,1].set_ylabel('deg C')

plt.tight_layout()
plt.show()
```
