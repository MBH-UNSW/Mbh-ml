```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# loading the generated synthetic data
# lowk have to change file location sometimes
df = pd.read_csv('tah_synthetic_healthy.csv')
df.head()

print(df.columns.tolist())
```

```python
t = df['timestamp']
left_flow = df['left_flow_lpm']
right_flow = df['right_flow_lpm']
left_inlet = df['left_inlet_pressure_mmHg']
right_inlet = df['right_inlet_pressure_mmHg']
left_upstream = df['left_upstream_pressure_mmHg']
left_downstream = df['left_downstream_pressure_mmHg']
right_upstream = df['right_upstream_pressure_mmHg']
right_downstream = df['right_downstream_pressure_mmHg']
pump_rpm = df['pump_rpm']
temp = df['tah_temperature_c']
```

```python
# only plotting first 1000 points otherwise its just a solid block
n = 1000

fig, axes = plt.subplots(4, 3, figsize=(15, 11))
axes = axes.flatten()

cols = [
    ('left_flow_lpm', left_flow, 'L/min'),
    ('right_flow_lpm', right_flow, 'L/min'),
    ('left_inlet_pressure_mmHg', left_inlet, 'mmHg'),
    ('right_inlet_pressure_mmHg', right_inlet, 'mmHg'),
    ('left_upstream_pressure_mmHg', left_upstream, 'mmHg'),
    ('left_downstream_pressure_mmHg', left_downstream, 'mmHg'),
    ('right_upstream_pressure_mmHg', right_upstream, 'mmHg'),
    ('right_downstream_pressure_mmHg', right_downstream, 'mmHg'),
    ('pump_rpm', pump_rpm, 'rpm'),
    ('tah_temperature_c', temp, 'deg C'),
]

for i, (name, series, unit) in enumerate(cols):
    axes[i].plot(t[:n], series[:n])
    axes[i].set_title(name)
    axes[i].set_ylabel(unit)

# hide the 2 unused subplot slots since we only have 10 signals
axes[10].axis('off')
axes[11].axis('off')

plt.tight_layout()
plt.show()
```
