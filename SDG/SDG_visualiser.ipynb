```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tah_synthetic_healthy.csv')

t = df['timestamp']


n = 1000
```

```python
cols = [
    'left_flow_lpm', 'right_flow_lpm', 'left_inlet_pressure_mmHg', 'right_inlet_pressure_mmHg', 'left_upstream_pressure_mmHg', 'left_downstream_pressure_mmHg', 
    'right_upstream_pressure_mmHg', 'right_downstream_pressure_mmHg', 'pump_rpm', 'tah_temperature_c'
]
```

```python
fig, axes = plt.subplots(4, 3, figsize=(15, 11))
axes = axes.flatten()

for i, col in enumerate(cols):
    axes[i].plot(t[:n], df[col][:n])
    axes[i].set_title(col)
```

```python
plt.tight_layout()
plt.show()

# stats to check values are in expected ranges
df.drop(columns='timestamp').describe().round(3)
```
